Create a class Circle with the constructor intializing the radius r,another function area to calculate the area of the circle, another function circumference to claculate the circumference of the (2pir)of the circle.

```python
class Circle:
  def __init__(self,r):
   self.radius = r
  def area(self):
    return 3.14*self.radius*self.radius
  def circumference(self):
    return 2*3.14*self.radius
c1 = Circle(4)
print(c1.area())
print(c1.circumference())
```

There are two classess namely num1 and num2 . In num1 class there are 2 variables a and b that needs to be initialized using constructor it also have a member fn show value to print/display this values. num2 class is inherited from num1 and it has 2 fns .1st one is show sum that display the sum of variables. 2nd one is show product that display the product of the variables

```python
class num1:
    def __init__(self,a,b):
       self.a=a
       self.b=b
    def showvalue(self):
       print(f"a={self.a}")
       print(f"b ={self.b}")
class num2(num1):
    def __init__(self,a,b):
       super().__init__(a,b)
    def showsum(self):
       print(f"sum= {self.a+self.b}")
    def showproduct(self):
       print(f"product={self.a*self.b}")


n1=num2(30,70)
n1.showvalue()
n1.showsum()
n1.showproduct()
```

There are 2 classess student and master student student is the base class and master student is the derived class .The class student has 2 member variables id and mark.there is a constructor to initialize this variable and a fn show result which will check the mark for passed and failed .if the mark is grester than or equal to 40 it is a pass otherwise it is a failed .similary,in the class master student contain a fn show result that will check whether the mark is greater than or equal to 50.if that is a case then it is a past otherwise it will be a fail.Intialize the student object and a master student object and call the functions respectively

```python
class student:
   def __init__(self,id,mark):
      self.id = id
      self.mark =mark
   def show_result(self):
    if self.mark>=40:
      print("pass")
    else:
      print("Fail")

class master_student(student):
   def __init__(self,id,mark):
    super().__init__(id,mark)
    if self.mark>=50:
      print("pass")
    else:
      print("Fail")

s1=student(101,50)
s2=master_student(66,75)
s1.show_result()
s2.show_result()
```
